```python
"""2) Scrivere una classe Rettagolo con gli attributi lunghezza e larghezza, e i metodi area e perimeter
che restituiscono rispettivamente l'area e il perimetro del rettangolo."""

class Rettangolo:
    def __init__(self,lunghezza:float,larghezza:float) -> None:
        self.luinghezza = lunghezza
        self.larghezza = larghezza

    def area(self) -> float:
       return self.larghezza * self.luinghezza
    
    def perimeter(self) -> float:
        return (self.luinghezza + self.larghezza) * 2
    
r1 = Rettangolo(10.34,43.6)   
print(f"area rettangolo: {r1.area()} e perimetro rettangolo: {r1.perimeter()}") 
```

```python
"""8) Definire una classe ContoBancario con l'attributo saldo e due metodi deposita e preleva che
aggiungono o sottraggono una certa quantità dal saldo del conto."""

class ContoBancario:

    def  __init__(self,nome:str,saldo:float = 0):
        self._saldo = saldo
        self.nome = nome

    @property
    def saldo(self):
        return self._saldo    
    
    #@property
    #def nome(self):
    #    return self.nome

    
    @saldo.setter
    def saldo(self,valore):
        self._saldo = valore


    def print_saldo(self):
        print(f"conto {self.nome} {self.saldo}")

    def deposita(self,amount:float):
        try:
           if amount < 0:
               raise Exception("errore  somma da depositare")
           else:
               sold = self.saldo
               sold += amount
               self.saldo = sold 
               self.print_saldo()
        except Exception as e:
            print(e)


    def preleva(self,amount:float) -> None:
        try:
            if amount <= 0:
                raise Exception("errore somma da prelevare") 
            else:
               sold = self.saldo
               sold -= amount
               #print(sold)
               self.saldo = sold 
               self.print_saldo()  
        except Exception as e:
            print(e)


c1 = ContoBancario("ange")
c2 = ContoBancario("bonheur",500)  

c1.preleva(0)
c2.preleva(100)

c1.deposita(1000000)
c2.deposita(-10)

            
                  

```

```python
"""9) Definire una classe Punto con gli attributi x e y e un metodo distanza_dall_origine che calcola la
distanza del punto dall'origine (0, 0)."""

import math

class Punto:
    def __init__(self,x,y):
        self._x = x
        self._y = y

    @property #getter
    def x(self):
      return self._x
    
    @property #getter
    def y(self):
      return self._y
    
    @x.setter#setter
    def x(self,val):
       self._x = val

    @y.setter#setter
    def y(self,val):
       self._y = val  

    def __str__(self):
       return f"""x = {self.x}\ny = {self.y}"""   

    def  distanza_dall_origine(self):
       d =  math.sqrt((self.x *self.x)+(self.y * self.y))
       return d

p1 = Punto(5,6)
p2 = Punto(2,3)

print(p1)
print(p2)

print(p1.distanza_dall_origine())
print(p2.distanza_dall_origine())
    
    

 

```

```python
"""16) Creare una classe NumeroComplesso con gli attributi parte_reale e parte_immaginaria. Definire
il metodo magico add per sommare due numeri complessi."""

class NumeroComplesso:
    def __init__(self,parte_reale:int = 0, parte_immaginaria:int = 0 ):
        self.parte_immaginaria =  parte_immaginaria
        self.parte_reale = parte_reale

    def __add__(self,other):
        return  NumeroComplesso(self.parte_reale + other.parte_reale, self.parte_immaginaria + other.parte_immaginaria)

    def __str__(self):
        return f"{self.parte_reale} + i{self.parte_immaginaria}"  
    
n1 =   NumeroComplesso(2,4)
n2 = NumeroComplesso() 
n3 = n1 + n2

print(n1)
print(n2)
print(n3)
```

```python
"""18) Creare una classe Prodotto con gli attributi nome e prezzo. Definire il metodo magico eq per
confrontare due prodotti in base al nome e al prezzo."""

class Prodotto:
    def __init__(self,nome:str = "", prezzo:float = 0):
        self.nome = nome
        self.prezzo = prezzo

    def __eq__(self, other):
        return self.nome == other.nome and self.prezzo == other.prezzo

    def __str__(self):
        return f"prodotto:{self.nome} -------> prezzo: {self.prezzo}"  

lista_prodotti = [Prodotto("guanciale",100),Prodotto(),Prodotto("pane"),Prodotto(90)]    

for i in lista_prodotti:
    print(i)
```

```python
"""20) Creare una classe ListaIntelligente con l'attributo lista. Definire il metodo magico getitem per
accedere agli elementi della lista tramite l'indice."""

class ListaIntelligente:
    def __init__(self, lista:list):
        self.lista = lista

    def __getitem__ (self,indice):
        return self.lista[indice]  

lista = [2,45,67,90]
l = ListaIntelligente(lista)

print(l[3])

```
