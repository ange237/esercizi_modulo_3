```python
"""2) Creare una classe ContoInBanca che incapsuli un attributo saldo e abbia metodi per depositare e
prelevare denaro dal conto. Dovremmo anche assicurarci che il saldo non possa essere negativo."""

class ContoInBanca ():
    def __init__(self,nome_titolare:str,saldo:float = 0) -> None:
        self.__saldo = saldo
        self.__nome_titolare = nome_titolare

    def get_saldo(self) -> float:
        return self.__saldo 

    def set_saldo(self,val:float) -> float:
        if self.__saldo < 0:
            raise Exception("saldo inferiore a zero")
        elif val < 0:
            raise Exception("saldo inferiore a zero")
        self.__saldo = val   
        


    def get_nome_titolare(self) -> str:
        return self.__nome_titolare    
    
    def set_nome_titolare(self,val:str) -> None:
        self.__nome_titolare = val

    def deposita(self,val:float):
        try:
         if val < 0:
             raise Exception("il valore da depositare non deve essere minore di zero")
         else:
            self.set_saldo(self.get_saldo() + val) 
            print("operazione conclusa con successo")
        except Exception as e:
            print(e)    

    def preleva(self,val:float):
        try:
            if val < 0:
                raise Exception("il valore da prelevare non deve essere minore di zero")
            else:
                self.set_saldo(self.get_saldo() - val)
                print("operazione conclusa con successo")   
        except Exception as e:
            print(e)

    def __str__(self):
        return f"{self.get_nome_titolare()} : ----->  {self.get_saldo()}"        


c1 = ContoInBanca("ange")
c2 = ContoInBanca("bonheur",10000)

c1.preleva(100)
print(c1)
print("======================================")
c2.deposita(-500)
print(c2)
print("======================================")
c1.deposita(900)
print(c1)
print("======================================")
c2.preleva(50000)
print(c2)
```
